```python
import os
import requests
import pandas as pd
from tqdm import tqdm

# === CONFIGURATION ===
CSV_SOURCE = "/Users/yanndanneels-coignard/Documents/definitifvins/SITEVINS/data/final/base_vin_final.csv"       # CSV d'entrée
CSV_OUTPUT = "/Users/yanndanneels-coignard/Documents/definitifvins/SITEVINS/data/final/URLimages.csv"            # CSV de sortie
IMG_DIR = "/Users/yanndanneels-coignard/Documents/definitifvins/SITEVINS/data/final/images"                      # Dossier où enregistrer les images
BASE_URL = "https://www.vinatis.com/"

# Créer le dossier d’images s’il n’existe pas
os.makedirs(IMG_DIR, exist_ok=True)

# Charger les données du CSV
df = pd.read_csv(CSV_SOURCE)

# Conserver les colonnes utiles et supprimer les lignes sans 'visuel'
df = df[["Unnamed: 0", "visuel"]].dropna()
df = df.rename(columns={"Unnamed: 0": "ID"})

# Construire l’URL complète et le nom de fichier PNG
df["URL"] = BASE_URL + df["visuel"].str.lstrip("/")
df["NomFichier"] = df["ID"].apply(lambda x: f"{int(x):05d}.png")

# Liste des téléchargements réussis
resultats = []

# Télécharger chaque image
for _, row in tqdm(df.iterrows(), total=len(df)):
    url = row["URL"]
    nom_fichier = row["NomFichier"]
    id_ = row["ID"]
    chemin_image = os.path.join(IMG_DIR, nom_fichier)

    try:
        r = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}, timeout=10)
        r.raise_for_status()
        with open(chemin_image, "wb") as f:
            f.write(r.content)
        resultats.append({"ID": id_, "URL": url, "NomFichier": nom_fichier})
    except Exception as e:
        print(f"❌ Erreur ID {id_} : {e}")

# Créer un nouveau DataFrame avec les images téléchargées
df_result = pd.DataFrame(resultats)

# Enregistrer dans un nouveau fichier CSV
df_result.to_csv(CSV_OUTPUT, index=False)

print(f"✅ {len(df_result)} images téléchargées avec succès.")
print(f"📁 CSV généré : {CSV_OUTPUT}")
os.makedirs(IMG_DIR, exist_ok=True)

# Charger les données du CSV
df = pd.read_csv(CSV_SOURCE)

# Conserver les colonnes utiles et supprimer les lignes sans 'visuel'
df = df[["Unnamed: 0", "visuel"]].dropna()
df = df.rename(columns={"Unnamed: 0": "ID"})

# Construire l’URL complète et le nom de fichier PNG
df["URL"] = BASE_URL + df["visuel"].str.lstrip("/")
df["NomFichier"] = df["ID"].apply(lambda x: f"{int(x):05d}.png")

# Liste des téléchargements réussis
resultats = []

# Télécharger chaque image
for _, row in tqdm(df.iterrows(), total=len(df)):
    url = row["URL"]
    nom_fichier = row["NomFichier"]
    id_ = row["ID"]
    chemin_image = os.path.join(IMG_DIR, nom_fichier)

    try:
        r = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}, timeout=10)
        r.raise_for_status()
        with open(chemin_image, "wb") as f:
            f.write(r.content)
        resultats.append({"ID": id_, "URL": url, "NomFichier": nom_fichier})
    except Exception as e:
        print(f"❌ Erreur ID {id_} : {e}")

# Créer un nouveau DataFrame avec les images téléchargées
df_result = pd.DataFrame(resultats)

# Enregistrer dans un nouveau fichier CSV
df_result.to_csv(CSV_OUTPUT, index=False)

print(f"✅ {len(df_result)} images téléchargées avec succès.")
print(f"📁 CSV généré : {CSV_OUTPUT}")
```
