Explication
colonnes_racines : liste les “racines” de tes variables one-hot (ex : “couleur_”).

Pour chaque racine : on trouve la colonne one-hot activée pour le vin cible (celle qui vaut 1). On applique un poids fort juste sur cette colonne dans la distance.

Le reste a un poids de 1.

Le prix (colonne numérique), tu lui mets son poids séparément.

Résultat : le KNN va vraiment privilégier les vins de même couleur, même type, même pays…
C’est exactement ce que tu veux !



```python
def recommander_vins_affichage_simple(
    df, 
    df_num, 
    vin_cible_nom, 
    colonnes_racines=('couleur_', 'type_', 'pays_','millesime_'),
    poids=(200, 200, 3, 200),
    colonne_prix='prix_std',
    poids_prix=100,
    n_voisins=5
):
    if vin_cible_nom not in df['nom'].values:
        raise ValueError(f"Vin '{vin_cible_nom}' non trouvé !")
    index_cible = df[df['nom'] == vin_cible_nom].index[0]

    colonnes_num = list(df_num.columns)
    poids_vect = np.ones(len(colonnes_num))
    for racine, p in zip(colonnes_racines, poids):
        for col in [col for col in colonnes_num if col.startswith(racine)]:
            poids_vect[colonnes_num.index(col)] = p
    if colonne_prix in colonnes_num:
        poids_vect[colonnes_num.index(colonne_prix)] = poids_prix

    vecteur_cible = df_num.iloc[index_cible].values.astype(float)
    mat = df_num.values.astype(float)
    distances = np.sqrt(np.sum(((mat - vecteur_cible) * poids_vect) ** 2, axis=1))
    distances[index_cible] = np.inf

    indices_voisins = np.argsort(distances)[:n_voisins-1]
    df_voisins = df.iloc[indices_voisins].copy()
    df_voisins['distance'] = distances[indices_voisins]

    df_bio = df[(df['bio_bool'] == 1) & (~df.index.isin(indices_voisins)) & (df.index != index_cible)]
    if not df_bio.empty:
        bios_num = df_num.loc[df_bio.index]
        bios_distances = np.sqrt(np.sum(((bios_num.values - vecteur_cible) * poids_vect) ** 2, axis=1))
        idx_min = np.argmin(bios_distances)
        vin_bio_index = df_bio.index[idx_min]
        df_voisins = pd.concat([df_voisins, df.loc[[vin_bio_index]].assign(distance=bios_distances[idx_min])])

    df_voisins = df_voisins.sort_values('distance')

    colonnes_clefs = [ 'nom', 'couleur', 'pays', 'bio_bool', 'type_produit', 'prix','distance']
    colonnes_clefs_cible = [col for col in colonnes_clefs if col != 'distance']
    cible = df.loc[[index_cible], colonnes_clefs_cible].copy()
    cible['distance'] = 0.0
    cible['nom'] = cible['nom'].values[0] + " (CIBLE)"

    def format_bio(val):
        return "O" if val == 1 else "N"
    cible['bio_bool'] = cible['bio_bool'].apply(format_bio)
    df_voisins['bio_bool'] = df_voisins['bio_bool'].apply(format_bio)

    final = pd.concat([cible, df_voisins[colonnes_clefs]], ignore_index=True)
    final = final[colonnes_clefs]

    print("\n=== RECOMMANDATIONS VIN (voisinage pondéré) ===")
    print(final.to_string(index=False))
    return final
```

```python
vin_cible_nom = 'Sainte Croix - Rosé 2024'
result = recommander_vins_affichage_simple(df, df_num, vin_cible_nom)
```

df de recommandation 

```python
# DataFrame résultat vide
df_reco = pd.DataFrame(index=df['nom'], columns=['reco1', 'reco2', 'reco3', 'reco4'])

for vin_nom in df['nom']:
    try:
        # On prend 5 voisins (1 cible + 4 recos)
        result = recommander_vins_affichage_simple(df, df_num, vin_nom, n_voisins=5)
        noms_recos = result['nom'].tolist()[1:5]  # On saute la cible elle-même
        df_reco.loc[vin_nom] = noms_recos
    except Exception as e:
        df_reco.loc[vin_nom] = [None, None, None, None]

df_reco.index.name = 'vin_cible'
```
