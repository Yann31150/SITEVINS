```python
import pandas as pd
import re

```

```python
 df = pd.read_csv("base_vin_final_temps.csv")
 df_image = pd.read_csv("URLimages.csv")

```

```python
import re

def extract_id_visuel_url(url):
    if isinstance(url, str):
        match = re.search(r'(\d+)-', url)
        if match:
            return match.group(1)
    return None

df_image['id_visuel'] = df_image['URL'].apply(extract_id_visuel_url)
```

```python
import re

def extract_id_visuel(visuel):
    if isinstance(visuel, str):
        match = re.search(r'(\d+)-', visuel)
        if match:
            return match.group(1)
    return None

df['id_visuel'] = df['visuel'].apply(extract_id_visuel)
```

```python
print(df[['visuel', 'id_visuel']].head())
print(df_image[['URL', 'id_visuel']].head())
```

```python
df = df.merge(
    df_image[['id_visuel', 'NomFichier']],
    on='id_visuel',
    how='left'  # ou 'inner' si tu veux garder uniquement les correspondances
)
```
